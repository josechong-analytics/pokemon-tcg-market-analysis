```python
import pandas as pd
import numpy as np
from google.colab import files

# 1. Force file upload dialog box
print("Please upload your raw CSV file below:")
uploaded = files.upload()

# Get the uploaded filename dynamically
input_filename = list(uploaded.keys())[0]

def pokemon_cards_clean_100(input_csv):
    print("\n--- STARTING PROCESS PHASE (DATA CLEANING) ---")

    # Load data
    df = pd.read_csv(input_csv)
    print(f"Initial raw rows loaded: {len(df)}")

    # Remove exact duplicates based on the real identifying columns
    df = df.drop_duplicates(subset=['title', 'card_number'], keep='first')
    print(f"Rows remaining after removing duplicates: {len(df)}")

    # String normalization
    if 'pokemon_name' in df.columns:
        df['pokemon_name'] = df['pokemon_name'].astype(str).str.strip()
    if 'set_name' in df.columns:
        df['set_name'] = df['set_name'].astype(str).str.strip()
    if 'rarity_class' in df.columns:
        df['rarity_class'] = df['rarity_class'].astype(str).str.strip()

    # Financial column cleanup
    if 'price_usd' in df.columns:
        df['price_usd'] = df['price_usd'].astype(str).str.replace('$', '', regex=False)
        df['price_usd'] = pd.to_numeric(df['price_usd'], errors='coerce')

    # Date formatting
    if 'release_date' in df.columns:
        df['release_date'] = pd.to_datetime(df['release_date'], errors='coerce').dt.strftime('%Y-%m-%d')
    else:
        df['release_date'] = np.nan

    # Integrity checks
    if 'stock' in df.columns:
        df['stock'] = df['stock'].fillna(0).astype(int)
    else:
        df['stock'] = 0

    print("\n--- CLEANING SUMMARY ---")
    print(f"Missing release dates (structurally preserved): {df['release_date'].isna().sum()}")
    if 'price_usd' in df.columns:
        print(f"Average Unit Price: ${df['price_usd'].mean():.2f}")
    print(f"Total Physical Inventory Count: {df['stock'].sum()}")

    # Export clean dataset
    clean_filename = 'pokemon_cards_clean_100.csv'
    df.to_csv(clean_filename, index=False)

    try:
        files.download(clean_filename)
        print(f"\nClean dataset successfully exported and downloaded!")
    except Exception as e:
        print(f"\nFile created but automatic download blocked. Please download manually from the left folder.")

# Execution call using the dynamically detected filename
pokemon_cards_clean_100(input_filename)
```
